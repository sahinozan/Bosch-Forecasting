# LSTM on fictional demand

All data used here is independently generated and fictional.


```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'examples' / 'demand.csv').exists())
sys.path.insert(0, str(ROOT / 'code'))
import pandas as pd
from helper import load_series, split_series, forecast, evaluate
```

```python
train, test = split_series(load_series(ROOT / 'examples' / 'demand.csv'))
predicted = forecast(train, len(test), method='lstm', epochs=1)
evaluate(test, predicted)
```

```python
pd.DataFrame({'actual': test, 'predicted': predicted})
```
