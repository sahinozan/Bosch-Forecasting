# Export fictional demand to Excel

All data used here is independently generated and fictional.


```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'examples' / 'demand.csv').exists())
sys.path.insert(0, str(ROOT / 'code'))
import pandas as pd
from helper import load_series, split_series, forecast, evaluate
```

```python
data = pd.read_csv(ROOT / 'examples' / 'demand.csv', parse_dates=['date'])
wide = data.pivot(index='date', columns='item_id', values='quantity')
output = ROOT / 'outputs'
output.mkdir(exist_ok=True)
wide.to_excel(output / 'synthetic_demand.xlsx')
```

```python
from openpyxl import load_workbook
from openpyxl.packaging.core import DocumentProperties
book = load_workbook(output / 'synthetic_demand.xlsx')
book.properties = DocumentProperties(creator='Public demo', lastModifiedBy='Public demo')
book.save(output / 'synthetic_demand.xlsx')
```
